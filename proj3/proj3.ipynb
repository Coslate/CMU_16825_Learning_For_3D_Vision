## **Justification for the Results**

### **1. Voxel Grid Representation**
- Voxel-based reconstruction is limited by its discrete grid structure, which introduces artifacts and reduces reconstruction accuracy.
- The fixed resolution makes it difficult to capture fine details, especially for curved or intricate surfaces, leading to the most complicated model to train compared with point cloud and mesh reconstruction models.

### **2. Point Cloud Representation**
- Point cloud-based reconstruction achieves the highest F1 scores because it directly represents object surfaces without the constraints of a fixed grid or mesh topology.
- This flexibility allows it to retain fine geometric details, making it well-suited for tasks requiring high spatial accuracy, such as shape analysis and object recognition.

### **3. Mesh Representation**
- Mesh-based methods provide a structured representation of 3D surfaces, balancing flexibility and connectivity constraints.
- While they offer smooth, continuous surfaces, their accuracy depends on the quality of the initial template shape, like the icosphere, and the ability to deform it to match the target shap.
- The balence between the flexibility and the intitial template shape constraint gives the mesh-based reconstruction a moderate F1 scores, higher than voxel grid reconstruction , but lower than point cloud reconstruction, meaning that it has higher flexibility than voxel grid, but has more constraints than point cloud.
