### Trade-off of View Dependence
Introducing view dependence in NeRF improves realism by accurately modeling specular highlights and reflective effects, making objects with shiny surfaces appear more natural. However, this comes at the cost of higher model complexity, longer training times, and increased risk of overfitting to specific lighting conditions from the training dataset. While it significantly benefits reflective surfaces, its impact on diffuse objects is minimal, meaning that for certain scenes (like the Lego scene), the added computational cost may not be justified. To get the best results, the model should be designed carefully so it doesn't just memorize the training images.

In this implementation, I used the sphere tracing algorithm to render a torus by iteratively marching along ray directions based on the Signed Distance Function (SDF). The algorithm initializes points at ray origins and iteratively updates their positions by moving them along the ray direction by a step size equal to the SDF value at each point. If the SDF value is below a small threshold (self.epsilon=1e-7), the ray is considered to have converged, and its position is recorded. A boolean mask is maintained to track which rays intersected the surface. The algorithm stops either when all rays have converged or when the maximum number of iterations is reached. This adaptive step size ensures efficient convergence while preventing unnecessary computations. The rendered torus closely matches the TA's result, confirming the correctness of the implementation and demonstrating an effective approach to implicit surface rendering.

### MLP for Neural Surface Representation
The Multi-Layer Perceptron (MLP) is used to learn an implicit Signed Distance Function (SDF) for surface reconstruction. It takes 3D point coordinates (x, y, z) as input and predicts the signed distance to the nearest surface. The MLP consists of multiple fully connected layers (Linear layers) with ReLU activations and an optional positional encoding to enhance high-frequency details. A skip connection is introduced at layer 3 to help the network learn long-range dependencies. The final layer outputs a single signed distance value without an activation function, allowing it to be both positive and negative.

### Eikonal Loss
The Eikonal loss enforces the SDF property ||∇Φ(x)|| = 1, ensuring a smooth and well-behaved distance field. It is computed as the mean squared deviation of the gradient L2-norm from 1. This regularization prevents gradient exploding or vanishing issues and helps in reconstructing a more accurate surface. The loss is particularly important for learning well-defined level sets of the SDF, ensuring that the implicit surface remains sharp and physically meaningful.

### Observations from the Results:
1. Alpha (α) - A scaling factor that controls the overall magnitude of the density function:

Higher α → Sharper surface transition between empty space and the object.

Lower α → More gradual transition, leading to a softer surface.

From the results, α = 15 gives the best balance. Too high (e.g., α = 50, 100) leads to artifacts or loss of details. Too low (e.g., α = 0.1, 1) results in diffused, blurry and unclear surfaces.

2. Beta (β) - Determines how smoothly the density transitions around the zero level set of the SDF:

Higher β → Leads to excessive blurring in the color image.

Lower β → Produces a more distinct rendering.

From the results, β = 0.05 gives the best reconstruction, while high β (e.g., 0.5) causes severe blurring in color and loss of details in geometry.

### Answer of the Questions:

1. How does high beta bias your learned SDF? What about low beta?

High beta leads to a blurred transition from occupied to free space, meaning the density function is spread out over a larger region. This can make the rendered color looks blurry because the density does not sharply define a surface.

Low beta makes the transition sharper, forcing the SDF to define a sharply bounded surface. This is better for sharp and accurate geometry but may cause instability if set too low.

2. Would an SDF be easier to train with volume rendering and low beta or high beta? Why?

An SDF is easier to train with a moderate or higher beta, rather than a very low beta because higher beta results in smoother gradients, making optimization more stable, but if the beta is too high, it would make that training harder because the density is too spread out, making it difficult to accurately localize the surface.

If beta is too low, the SDF behaves almost like a binary indicator function, making the optimization harder. Specifically, low beta would make gradient-based optimization struggle with extremely sharp transition, as tiny updates may cause significant changes in the output.

A moderate beta (e.g., 0.05 in my experiments) ensures that the SDF still provides a well-defined surface while keeping training stable.

3. Would you be more likely to learn an accurate surface with high beta or low beta? Why?

It would be more likely to learn an accurate surface with low beta (but not too low), because low beta leads to a more accurate surface.

On the other hand, high beta results in soft density transitions, making the surface poorly defined and blurry, and thereby harder to be used with to learn an accurate surface. 

From my expereiments above, moderate beta (e.g., 0.05) provides a good balance between a well-defined surface and stable optimization.
