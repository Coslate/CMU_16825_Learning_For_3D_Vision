#### Discussion Tradeoff Between Rendering Mesh VS Point Cloud
![PerformanceCompare](./compare_mesh_point_coud.png)

As we can see from the above standard output, the Point Cloud Rendering has much smaller execution time (2.083 seconds vs 14.875 seconds) and slightly smaller GPU memory usage (124.37 MB vs 162.09 MB) compared with Mesh Rendering under same number of samples and voxel size (600). Let's analyze as following:

1. Point Cloud Rendering is faster than Mesh Rendering:

A point cloud renderer simply plots points onto the screen, while a mesh renderer must perform triangle rasterization, normal computation, and fragment shading, which is computationally much more expensive. Besides, the mesh renderer needs to handle vertex connectivity and fill pixels inside triangles. On the other hand, the point cloud renderer does not need to handle triangle interpolation or normals calculation.

2. Point Cloud Rendering uses less GPU memory than Mesh Rendering:

A point cloud renderer only needs to store XYZ positions and colors, while a mesh renderer has to store triangle vertices, faces, textures, and normal vectors. In addition, the mesh renderer needs to calculate shading, lighting, and interpolation accross triangles, requiring additional GPU memory buffers.

3. Rendering Quality:

The rendering result of a point cloud renderer shows a smoother appearance, but it lacks surface connectivity, which could lead to gaps when the density (number of samples) is low. On the other hand, the mesh rendering provides well-defined structure with sharper edges and better shading, but it could become blocky if the resolution is low.

4. Ease of Use:

Point clouds are easier to generate, but they are harder to render and use in simulations due to the lack of surface continuity. On the other hand, meshes are more complex to generate because they require algorithms like marching cube to grab the zero level-set point of an implicit function, but they are easier to be used in physics simulations and graphics pipelines as they provide continuous, structured surfaces with well-defined connectivity between vertices.
