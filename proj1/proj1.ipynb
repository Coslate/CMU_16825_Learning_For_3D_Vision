## Task 4: Camera Transformations (10 points)

In the first transform:

$$
T_{\text{relative}} = \begin{bmatrix} 0 & 0 & 0 \end{bmatrix}
$$

$$
R_{\text{relative}} =
\begin{bmatrix}
0 & 1 & 0 \\
-1 & 0 & 0 \\
0 & 0 & 1
\end{bmatrix}
$$

rendered result:

![Rendered Result](./outputs/trans1_cow.jpg)

In the second transform:

$$
T_{\text{relative}} = \begin{bmatrix} 0 & 0 & 2 \end{bmatrix}
$$

$$
R_{\text{relative}} =
\begin{bmatrix}
1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}
$$ with \( \theta = 5^\circ \)

rendered result:

![Rendered Result](./outputs/trans2_cow.jpg)

In the third transform:

$$
T_{\text{relative}} = \begin{bmatrix} 0.35 & -0.5 & 0 \end{bmatrix}
$$

$$
R_{\text{relative3}} =
\begin{bmatrix}
\cos\left(\frac{\pi}{180} \times \theta\right) & 0 & \sin\left(\frac{\pi}{180} \times \theta\right) \\
0 & 1 & 0 \\
-\sin\left(\frac{\pi}{180} \times \theta\right) & 0 & \cos\left(\frac{\pi}{180} \times \theta\right)
\end{bmatrix}^T
\quad \text{with } \theta = -5^\circ
$$

rendered result:

![Rendered Result](./outputs/trans3_cow.jpg)

In the fourth transform:

In order to place the final camera position in [0, 0, -3], I set the following T_relative, with corresponding R_relative for the rotation on y-axis.
$$
T_{\text{relative}} = \begin{bmatrix} -3 & 0 & 3\end{bmatrix}
$$

$$
R_{\text{relative3}} =
\begin{bmatrix}
\cos\left(\frac{\pi}{180} \times \theta\right) & 0 & \sin\left(\frac{\pi}{180} \times \theta\right) \\
0 & 1 & 0 \\
-\sin\left(\frac{\pi}{180} \times \theta\right) & 0 & \cos\left(\frac{\pi}{180} \times \theta\right)
\end{bmatrix}^T
\quad \text{with } \theta = -90^\circ
$$
