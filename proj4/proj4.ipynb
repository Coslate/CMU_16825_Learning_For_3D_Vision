### Modifications Compared to Task 1.2.2 (Baseline)

#### 1. Switched from Isotropic to Anisotropic Gaussians
In the baseline, isotropic Gaussians were used, which only allow uniform scaling across all directions. This limits how well the Gaussians can represent complex surface shapes. In the improved version, anisotropic Gaussians are enabled by introducing quaternion-based orientation parameters. This significantly enhances the model’s expressiveness and reconstruction accuracy.

#### 2. Added SSIM Loss in Addition to L1 Loss
While L1 loss encourages pixel-wise similarity, it may not correlate well with human perception. SSIM (Structural Similarity Index) captures perceptual quality by considering luminance, contrast, and structure. Including SSIM helps the model preserve global and local visual features, especially textures. This makes the final renderings sharper and more natural.

#### 3. Introduced Learnable Weight for SSIM Loss (`lambda_ssim`)
Instead of using a fixed scalar for balancing SSIM loss, a learnable `lambda_ssim` was added. This gives the model the ability to automatically adjust the importance of SSIM relative to L1 during training. It reduces the need for manual hyperparameter tuning. The network adapts more effectively as it learns what’s important to optimize.

#### 4. Adjusted Learning Rates for Different Gaussian Parameters
Different Gaussian attributes such as `means`, `colours`, `pre_act_scales`, and `pre_act_opacities` were assigned tailored learning rates. This prevents unstable updates for sensitive parameters while allowing others to learn faster. For instance, colors were updated slowly to retain visual stability, while means had moderate learning speed for spatial accuracy. This fine-grained control boosts training stability and convergence quality.

#### 5. Included a New Learnable Parameter: `pre_act_quats`
The quaternion (`pre_act_quats`) parameter controls the orientation of anisotropic Gaussians. It was not present in the isotropic baseline but is crucial when modeling anisotropy. Adding it makes the representation significantly more flexible in 3D space. A small learning rate was used to stabilize its update, considering its high sensitivity.

#### 6. Assigned Separable Cosine Annealing Schedulers to Each Parameter Group
Each learnable parameter group (e.g., `pre_act_scales`, `means`, `pre_act_quats`, `colours`, `pre_act_opacities`) was assigned its own learning rate scheduler using cosine annealing or step decay. This allows precise control over how each component evolves over training. For instance, quats update gradually with a long schedule to preserve rotational smoothness. This per-parameter scheduling strategy improves both convergence and final quality.

#### 7. Increased Learning Rates Compared to Baseline
For each parameter group (e.g., `pre_act_opacities`, `pre_act_scales`, `pre_act_quats`, `means`, `colours`), I increased the learning rates relative to those used in the baseline setup. This helps the model converge faster in the early stages, particularly on the harder dataset. Higher learning rates allow the optimizer to make more aggressive progress initially. However, this is carefully managed with separate schedulers to avoid instability.
