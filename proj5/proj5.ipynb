### Interpretation

The segmentation results demonstrate strong performance on common object types like chairs, with good cases achieving over 99% accuracy and accurately distinguishing parts such as seats, backs, and legs. These examples show that the model can effectively generalize when the object shape is regular and well-represented in the training data. However, in the bad cases, the model struggles with less common geometries or noisy point clouds, leading to significant drops in accuracy (around 43–48%). In bad case 1, the prediction significantly misclassifies the seat area (originally red parts at the bottom of the ground truth) as part of the blue base, indicating difficulty in preserving clear boundaries between sofa components that are stacked and visually blended in point cloud space. In bad case 2, the model heavily confuses the magenta headrest and yellow side arm with the cyan body and red seat, collapsing multiple distinct parts into fewer predicted labels. This suggests challenges in segmenting objects with subtle geometric variation and symmetry, especially when parts lack clear spatial separation. The failure cases indicate that to improve the model, issues related to part overlap and misalignment need to be effectively addressed.
